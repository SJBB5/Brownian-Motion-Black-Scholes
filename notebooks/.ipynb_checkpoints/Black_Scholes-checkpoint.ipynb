We covered GBM using the equation,

$$dS_t = \mu S_t dt + \sigma S_t dW_t$$

The famous black scholes equation is

$$
\frac{\partial V}{\partial t}
+\frac{1}{2}\sigma^{2} S^{2} \frac{\partial^{2} V}{\partial S^{2}}
+r S \frac{\partial V}{\partial S}
- r V = 0,
\qquad
V(S,T)=\max(S-K,0).
$$

and works under the assumptions that stocks follow Geometric Brownian Motion.

```python
import numpy as np
from scipy.stats import norm
from src.funcs import blackScholes, build_bs_param_df

df = build_bs_param_df("^SPX", expiration_index=5, r=0.04, opt_type="C")
```

```python
df.head()
```

```python
row = df.iloc[0]

bs_price = blackScholes(
    r=row["r"],
    S=row["S"],
    K=row["K"],
    T=row["T"],
    sigma=row["sigma"],
    type=row["type"]
)

print(bs_price)
```
