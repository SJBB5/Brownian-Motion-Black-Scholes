```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# Parameters
S0 = 100
K  = 110
T  = 1.0       
r  = 0.05      
sigma = 0.25
N = 200000      

# simulate ST under GBM 
Z = np.random.randn(N)
ST = S0 * np.exp((r - 0.5*sigma**2)*T + sigma*np.sqrt(T)*Z)

# call payoff at expiration
payoff = np.maximum(ST - K, 0)

# Histogram
counts, bins, patches = plt.hist(ST, bins=120, density=True)

# Shade the region where S_T > K 
for left, right, patch in zip(bins[:-1], bins[1:], patches):
    if left >= K:
        patch.set_alpha(0.5)  

# line at strike
plt.axvline(K, linewidth=2)

plt.title(r"Simulated $S_T$ distribution with strike $K$ and call payoff region")
plt.xlabel(r"$S_T$")
plt.ylabel("Density")
plt.show()
```
